## Etapa 3: Respondendo as perguntas

1. Quantos pedidos foram feitos no total?

```python
# Como não temos registros duplicados, basta indicar o número de linhas
print(f"Foram feitos {len(df)} pedidos")
```

2. Quais são os restaurantes únicos que aparecem no dataset?

```python
# removi duplicatas e ordenei por ordem alfabética
df['Restaurant Name'].drop_duplicates().sort_values().reset_index(drop=True)
```

3. Quantos pedidos foram feitos em cada ano (2024 vs 2025)?

```python
# usando o value_counts e agrupando por ano
df['Order Time'].dt.year.value_counts().sort_index()
```

```python
# decidi também deixar registrado minha primeira forma de resolução da questão
dados_ano = df[df['Order Time'].dt.year==2025]

print(f'Pedidos em 2024: {len(df) - len(dados_ano)}\nPedidos em 2025: {len(dados_ano)}')
```

4. Qual foi a média de gorjeta (Tip ($)) por pedido?

```python
# utilizando .mean() para calcular a média
media_gorjeta = df['Tip ($)'].mean()
print(f"A média de gorjeta por pedido foi de ${media_gorjeta:.2f}")
```

5. Qual o pedido com o maior valor total (Price + Tip)?

```python
valor_pedido = df['Price ($)'] + df['Tip ($)']

print(f'O pedido de maior valor custou ${valor_pedido.max():.2f}')
```

6. Quantos pedidos foram feitos aos fins de semana? E durante a semana?

```python
# aproveitei para usar a operação de negação que vi aplicação recentemente
print(f"Pedidos feitos aos fins de semana: {df['Weekend Order'].sum()} pedidos")
print(f"Pedidos feitos durante a semana: {(~df['Weekend Order']).sum()} pedidos")
```

```python
# outra alternativa, criada com o ChatGPT
df['Weekend Order'].value_counts().rename(index={True: 'Fins de semana', False: 'Dias úteis'})
```

```python
weekend_order = df['Weekend Order']==True
print(f'Pedidos feitos aos fins de semana: {weekend_order.sum()}\
    \nPedidos feitos durante a semana: {len(df) - weekend_order.sum()}')
```

7. Qual restaurante teve o maior número de pedidos?

```python
df['Restaurant Name'].value_counts()
```

8. Qual é o tempo médio de entrega por restaurante?

```python
# utilizando o groupby para agrupar os restaurantes e calculando a media da coluna de duração. O sort_values mostra quem entrega mais rapido
df.groupby('Restaurant Name')['Delivery Duration (min)'].mean().round(2).sort_values()
```

9. Qual é o tamanho de taco mais vendido? E o tipo mais popular? 

```python
# usando a mesma lógica da 8 para mostrar o tamanho
df.groupby('Taco Size')['Order ID'].count().sort_values(ascending=False)
```

```python
df.groupby('Taco Type')['Order ID'].count().sort_values(ascending=False)
```
